## Task 4 Answers
- **Volume:** The file's size is 13,619 bytes(13.6 KB), it is not big data because the dataset comfortably fits in Excel or single laptop's memory(like my laptop) can process it reasonable time.
- **Velocity:** In a year it would accumulate 18,250 rows per year(50 application per day * 365 days). Excel's limit is about 1,048,576 rows, so it would take about 57 years to fill one sheet(1,048,576/18,250).
- **Variety:** A real bank loan file would also include scanned national ID photos for KYC(the bank's identity check), which is unstructured data, because they have no predefined structure and can't be stored in rows and columns.
- **Veracity:** `monthly_income_etb` has 13 missing values and `has_collateral` has 6 miising values, so some applicats' information is incomplete. There are also 5 duplicate rows, which is why `applicant_id` is not unique.
- **Value:** The bank could tighten its collateral poicy, for example by requiring collateral for larger loans or charging high interest rate to applicants who offer none. We can look Task 5 as evidence, applicants without collateral default is 26% while those with collateral is 12.1%.

```python
p_default = (df["defaulted"] == "Yes").mean()
print("P(Default):", p_default)
rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
lambda x: (x == "Yes").mean()
)
print(rate_by_collateral)
```

## Task 5 Answers
- P(Default) = 0.1934426(19.3%)
- P(Default | Has Collateral) = 0.120805(12.1%)
- P(Default | No Collateral) = 0.260000(26%)
- Yes the pattern from Unit 1 holdes. Applicants without collateral default more often, by 13.9%(26% - 12.1%) points, roughly twice the default rate of applicants with collateral.
